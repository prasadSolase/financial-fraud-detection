```python
import pandas as pd
from pathlib import Path

# Load all four datasets
df1 = pd.read_csv(r"../data/raw/synthetic_fraud_dataset1.csv")
df2 = pd.read_csv(r"../data/raw/Fraud Detection.csv")
df3 = pd.read_csv(r"../data/raw/financial_fraud_detection_dataset.csv")
df4 = pd.read_csv(r"../data/raw/credit_card_fraud_dataset.csv")

datasets = {
    "Synthetic Fraud Dataset": df1,
    "Fraud Detection Dataset": df2,
    "Financial Fraud Dataset": df3,
    "Credit Card Fraud Dataset": df4
}

for name, df in datasets.items():
    print(f"\n{'='*50}")
    print(name)
    print("="*50)
    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())
    print("\nFirst 5 Rows:")
    display(df.head())
```

```python
# Compare all four datasets

import pandas as pd

datasets = {
    "Synthetic Fraud Dataset": (df1, "Fraud_Label"),
    "Fraud Detection Dataset": (df2, "isFraud"),
    "Financial Fraud Dataset": (df3, "Fraudulent"),
    "Credit Card Fraud Dataset": (df4, "IsFraud")
}

comparison = []

for name, (df, target) in datasets.items():
    comparison.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum(),
        "Target Column": target,
        "Target Distribution": df[target].value_counts().to_dict()
    })

comparison_df = pd.DataFrame(comparison)

display(comparison_df)
```

```python
# Check fraud and non-fraud counts separately

for name, (df, target) in datasets.items():
    print(f"\n{'='*50}")
    print(name)
    print("="*50)

    print(df[target].value_counts(dropna=False))

    print("\nPercentage Distribution:")
    print(
        (df[target].value_counts(normalize=True, dropna=False) * 100)
        .round(2)
    )
```
