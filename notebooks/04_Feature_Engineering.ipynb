```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv(r"../data/processed/credit_card_fraud_cleaned.csv")
```

```python
df
```

```python
# Convert TransactionDate to datetime format
df["TransactionDate"] = pd.to_datetime(
    df["TransactionDate"],
    errors="coerce"
)

# Check the data type
print(df["TransactionDate"].dtype)
```

```python
model_df = df.copy()

# Extract useful date features
model_df["TransactionMonth"] = model_df["TransactionDate"].dt.month
model_df["TransactionDay"] = model_df["TransactionDate"].dt.day
model_df["TransactionDayOfWeek"] = model_df["TransactionDate"].dt.dayofweek

# Display the new columns
display(
    model_df[
        [
            "TransactionDate",
            "TransactionMonth",
            "TransactionDay",
            "TransactionDayOfWeek"
        ]
    ].head()
)
```

```python
# Select candidate features
feature_columns = [
    "Amount",
    "MerchantID",
    "TransactionType",
    "Location",
    "TransactionMonth",
    "TransactionDay",
    "TransactionDayOfWeek"
]

X = model_df[feature_columns]
y = model_df["IsFraud"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature data types:")
print(X.dtypes)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
print("Unique Transaction IDs:", model_df["TransactionID"].nunique())
print("Total rows:", len(model_df))

print(
    "Transaction IDs are sequential:",
    model_df["TransactionID"].is_monotonic_increasing
)
```
